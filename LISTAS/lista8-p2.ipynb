# **Questão 18**

1 - Avaliações, Questionários, Histórico de pesquisas, métricas de tempo relacionadas a produtos e uso da plataforma

2 - Podem ser usados para formar a matriz usuario-item ou treinar modelos que serão usados para sistemas de recomendação

3 - Devido a coleta de dados de forma implícita, que pode acabar sendo sem permissão ou de forma invasiva, indo de enconto a leis de proteção digitais

```
import pandas as pd

ratings = pd.read_csv("ratings.csv")
movies = pd.read_csv("movies.csv")

ratings = pd.merge(movies, ratings)

movieRatings = ratings.pivot_table(
    index=['userid'],
    columns=['title'],
    values='rating'
)

starWarsRatings = movieRatings['Star Wars (1977)']

similarMovies = movieRatings.corrwith(starWarsRatings)
similarMovies = similarMovies.dropna()

display(similarMovies)
```

# **Questão 20**

## **Árvore de Decisão**

**Objetivo** - Gerar classificações para um conjunto de dados

**Vantagens** - O desempenho não é afetado por esacala dos dados e captura relações não lineares

**Desvantagens** - Propenso a overfitting e tende a ser inviesado pelo dataset

## **Random Forest**

**Objetivo** - Usar várias árvores de decisão para o mesmo problema e fazer previsões considerando a maioria

**Vantagens **- Menos inviesada e com menos chance de overfitting

**Desvantagens** - Complexidade e custo computacional, difícil de interpretar


## **Filtragem Colaborativa:**

**Objetivo** - Gerar recomendações com base em dados, avaliações, escolhas e comportamentos dos usuários e itens, usando padrões entre essas informações

**Vantagem** - Analisa com base em informações reais e experiências gerais dos usuários de tal serviço

**Desvantagem** - Problemas como o da sparsidade, ovelha negra, usuários e itens novos, aleḿ de que em alguns casos apresenta alto custo computacional ou de armazenamento(baseados em memória), E pode apresentar problemas como os custos da etapa de treinamento e o tempo necessário para isso(baseados em modelo)

## **Sistemas Baseados em Modelo:**

**Objetivo** - Filtragem colaborativa que usa um modelo treinado para fazer as previsões

**Vantagens** - Evita o custo de armazenamento de sistemas baseados em memória, respostas mais rápidas(evita varrer a matriz toda vez), mais escalabilidade

**Desvantagens** - O tempo, recursos e dados necessários para treinar o modelo; custo computacional e necessidade de 'retreinamento', o que é demorado e pode 'atrasar' como as previsões são feitas, pois uma avaliação de hoje será processada amanhã.

## **Sistemas Baseados em Memória**

Objetivo - Previsões por meio de calculos que varrem uma matriz de dados cruzados entre usuario-item

Vantagens - Maior simplicidade, atualização rápida

Desvantagens - Baixa escalabilidade, apresenta problemas de tempo e custo de armazenamento caso a matriz cresca muito, problema da sparsidade, ovleha negra e cold start(usuario/item novo)
