```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier

dataset = load_breast_cancer()
x = dataset.data
y = dataset.target

x_train, x_test, y_train, y_test = train_test_split(x, y, random_state=42)

scaler = StandardScaler()

train_x_normalizado = scaler.fit_transform(x_train)
test_x_normalizado = scaler.transform(x_test)

display(train_x_normalizado)
display(test_x_normalizado)

knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(train_x_normalizado, y_train)
previsao_knn = knn.predict(test_x_normalizado)

```

```python
matriz_confusao = confusion_matrix(y_test, previsao_knn)
print(matriz_confusao)

plot_matriz = ConfusionMatrixDisplay(matriz_confusao, display_labels=['Maligno / 0', 'Benigno / 1'])
plot_matriz.plot()
```
