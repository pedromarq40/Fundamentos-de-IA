```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_blobs
from sklearn.cluster import KMeans

#Criando dados falsos 
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)
display(X)

#instânciando o k_means
kmeans = KMeans(n_clusters=4, random_state=42, n_init='auto')

previsoes = kmeans.fit_predict(X)
display(previsoes)

#ENCONTRANDO OS CENTROÍDES
centroides = kmeans.cluster_centers_
display(centroides)

#Plotando o gráfico 
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=previsoes, s=50, cmap='viridis', alpha=0.8)
plt.scatter(centroides[:, 0], centroides[:, 1], c='red', s=200, marker='X', label='Centróides')

plt.title("Visualização de Clusters com K-Means")
plt.xlabel("Característica 1")
plt.ylabel("Característica 2")
plt.legend()
plt.show()
```

```python
X, _ = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)

inercia = []
valores_k = range(1, 11)

for k in valores_k:

    kmeans = KMeans(n_clusters=k, random_state=42, n_init='auto')
    kmeans.fit(X)
    inercia_k = kmeans.inertia_
    inercia.append(inercia_k)

#Plotando o Gráfico de Linha (O Método do Cotovelo)
plt.figure(figsize=(8, 5))
plt.plot(valores_k, inercia, marker='o', linestyle='--', color='b')

# Adicionando detalhes ao gráfico
plt.title('Método do Cotovelo para descobrir o K ideal')
plt.xlabel('Número de Clusters (K)')
plt.ylabel('Inércia')
plt.xticks(valores_k)
plt.grid(True)
plt.show()
```
